```python
campaign_perf_df = spark.table("googleads_silver.silver_campaign_performance")

ad_group_perf_df = spark.table("googleads_silver.silver_ad_group_performance")

ad_perf_df = spark.table("googleads_silver.silver_ad_performance")

keyword_perf_df = spark.table("googleads_silver.silver_keyword_performance")

search_term_df = spark.table("googleads_silver.silver_search_terms")

device_perf_df = spark.table("googleads_silver.silver_device_performance")

age_perf_df = spark.table("googleads_silver.silver_age_performance")

gender_perf_df = spark.table("googleads_silver.silver_gender_performance")

geo_perf_df = spark.table("googleads_silver.silver_geographic_performance")

landing_page_df = spark.table("googleads_silver.silver_landing_page_performance")

conversion_perf_df = spark.table("googleads_silver.silver_conversion_performance")

campaign_details_df = spark.table("googleads_silver.silver_campaign_details")

customer_df = spark.table("googleads_silver.silver_customer_info")

budget_df = spark.table("googleads_silver.silver_budget_information")

conversion_action_df = spark.table("googleads_silver.silver_conversion_actions")
```

```python
from pyspark.sql.functions import (
    col,
    sha2,
    concat_ws
)
```

```python
from pyspark.sql.functions import col, sha2

dim_campaign = (
    campaign_details_df.alias("c")
    .join(
        budget_df.alias("b"),
        col("c.campaign_budget_id") == col("b.campaign_budget_id"),
        "left"
    )
    .select(

        # Campaign surrogate key
        sha2(col("c.campaign_id"), 256).alias("campaign_pk"),

        # Budget surrogate key
        sha2(col("b.campaign_budget_id"), 256).alias("budget_pk"),

        # Natural key
        col("c.campaign_id"),

        # Attributes
        col("c.campaign_name"),
        col("c.campaign_status"),
        col("c.campaign_serving_status"),
        col("c.campaign_advertising_channel_type")
            .alias("advertising_channel_type"),
        col("c.campaign_bidding_strategy_type")
            .alias("bidding_strategy_type")

    )
    .distinct()
)

display(dim_campaign)
```

```python
dim_customer = customer_df.select(

    sha2(col("customer_id"),256).alias("customer_pk"),

    col("customer_id"),

    col("customer_currency_code").alias("currency_code"),

    col("customer_time_zone").alias("time_zone")

).distinct()

display(dim_customer)
```

```python
dim_budget = budget_df.select(

    sha2(col("campaign_budget_id"),256).alias("budget_pk"),

    col("campaign_budget_id").alias("budget_id"),

    col("campaign_budget_name").alias("budget_name"),

    col("budget_inr")

).distinct()

display(dim_budget)
```

```python
dim_conversion_action = conversion_action_df.select(

    sha2(col("conversion_action_id"),256)
    .alias("conversion_action_pk"),

    col("conversion_action_id"),

    col("conversion_action_name"),

    col("conversion_action_type_").alias("conversion_type"),

    col("conversion_action_status")

).distinct()

display(dim_conversion_action)
```

```python
dim_ad_group = ad_group_perf_df.select(

    "ad_group_id",
    "ad_group_name",
    "campaign_id"

).distinct()

dim_ad_group = dim_ad_group.join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "left"
).select(

    sha2(col("ad_group_id"),256).alias("ad_group_pk"),

    col("ad_group_id"),

    col("campaign_pk"),

    col("ad_group_name")

)

display(dim_ad_group)
```

```python
"""
Creates ad dimension table.

Source: silver_ad_performance
Grain: One row per ad

CHANGES:
- Uses ad_performance table because no ad metadata table exists.
- Added SHA256 surrogate key (ad_pk).
"""

dim_ad = ad_perf_df.select(

    # --- Surrogate Key (Primary Key) ---
    sha2(
        concat_ws(
            "||",
            col("ad_id"),
            col("ad_group_id")
        ),
        256
    ).alias("ad_pk"),

    # --- Natural Keys ---
    col("ad_id"),

    col("ad_group_id"),

    col("ad_group_name"),

    col("campaign_id"),

    col("campaign_name")

).distinct()

# Display sample
print("dim_ad sample:")
display(dim_ad)

print(f"\nCreated dim_ad with {dim_ad.count()} ads")

print(
    "Columns: ad_pk (SHA256), "
    "ad_id, "
    "ad_group_id, "
    "ad_group_name, "
    "campaign_id, "
    "campaign_name"
)
```

```python
dim_keyword = keyword_perf_df.select(

    col("keyword"),

    col("keyword_match_type")

).distinct()

dim_keyword = dim_keyword.withColumn(

    "keyword_pk",

    sha2(
        concat_ws(
            "||",
            col("keyword"),
            col("keyword_match_type")
        ),
        256
    )
)

display(dim_keyword)
```

```python
dim_search_term = search_term_df.select(

    sha2(col("search_term"),256)
    .alias("search_term_pk"),

    col("search_term")

).distinct()

display(dim_search_term)
```

```python
dim_device = device_perf_df.select(
    "device"
).distinct()

dim_device = dim_device.select(

    sha2(col("device"),256)
    .alias("device_pk"),

    col("device")
)

display(dim_device)
```

```python
dim_gender = gender_perf_df.select(
    "gender"
).distinct()

dim_gender = dim_gender.select(

    sha2(col("gender"),256)
    .alias("gender_pk"),

    col("gender")
)

display(dim_gender)
```

```python
dim_age_range = age_perf_df.select(
    "age_range"
).distinct()

dim_age_range = dim_age_range.select(

    sha2(col("age_range"),256)
    .alias("age_range_pk"),

    col("age_range")
)

display(dim_age_range)
```

```python
dim_location = geo_perf_df.select(
    "country_criterion_id"
).distinct()

dim_location = dim_location.select(

    sha2(col("country_criterion_id"),256)
    .alias("location_pk"),

    col("country_criterion_id")

)

display(dim_location)
```

```python
dim_landing_page = landing_page_df.select(
    "landing_page_url"
).distinct()

dim_landing_page = dim_landing_page.select(

    sha2(col("landing_page_url"),256)
    .alias("landing_page_pk"),

    col("landing_page_url")

)

display(dim_landing_page)
```

```python
"""
Creates campaign performance fact table.

Source: Silver campaign_performance
Grain: One row per (date, campaign)
"""

fact_campaign_perf = campaign_perf_df.alias("c").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("campaign_pk"),

    # Metrics
    col("c.metrics_impressions").alias("impressions"),
    col("c.metrics_clicks").alias("clicks"),
    col("c.metrics_conversions").alias("conversions"),
    col("c.cost_inr"),
    col("c.ctr_pct"),
    col("c.cpc_inr"),
    col("c.cvr_pct"),
    col("c.cpa_inr")

)

# Display sample
print("fact_campaign_perf sample:")
display(fact_campaign_perf)

print(f"\nCreated fact_campaign_perf with {fact_campaign_perf.count()} daily campaign records")
print("Columns: segments_date, campaign_pk, impressions, clicks, conversions, cost_inr, ctr_pct, cpc_inr, cvr_pct, cpa_inr")
```

```python
"""
Creates keyword performance fact table.

Source: Silver keyword_performance
Grain: One row per (date, campaign, ad_group, keyword)
"""

fact_keyword_perf = keyword_perf_df.alias("k").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_ad_group.select(
        "ad_group_id",
        "ad_group_pk"
    ),
    "ad_group_id",
    "inner"
).join(
    dim_keyword.select(
        "keyword",
        "keyword_match_type",
        "keyword_pk"
    ),
    ["keyword","keyword_match_type"],
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("campaign_pk"),
    col("ad_group_pk"),
    col("keyword_pk"),

    # Metrics
    col("k.metrics_impressions").alias("impressions"),
    col("k.metrics_clicks").alias("clicks"),
    col("k.metrics_conversions").alias("conversions"),
    col("k.cost_inr"),
    col("k.ctr_pct"),
    col("k.cpc_inr"),
    col("k.cvr_pct"),
    col("k.cpa_inr")

)

print("fact_keyword_perf sample:")
display(fact_keyword_perf)

print(f"\nCreated fact_keyword_perf with {fact_keyword_perf.count()} daily keyword records")
```

```python
"""
Creates search term performance fact table.

Source: Silver search_terms
Grain: One row per (date, campaign, ad_group, search_term)
"""

fact_search_term_perf = search_term_df.alias("st").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_ad_group.select(
        "ad_group_id",
        "ad_group_pk"
    ),
    "ad_group_id",
    "inner"
).join(
    dim_search_term.select(
        "search_term",
        "search_term_pk"
    ),
    "search_term",
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("campaign_pk"),
    col("ad_group_pk"),
    col("search_term_pk"),

    # Metrics
    col("st.metrics_impressions").alias("impressions"),
    col("st.metrics_clicks").alias("clicks"),
    col("st.metrics_conversions").alias("conversions"),
    col("st.cost_inr"),
    col("st.ctr_pct"),
    col("st.cpc_inr"),
    col("st.cvr_pct"),
    col("st.cpa_inr")

)

print("fact_search_term_perf sample:")
display(fact_search_term_perf)

print(f"\nCreated fact_search_term_perf with {fact_search_term_perf.count()} daily search term records")
```

```python
"""
Creates device performance fact table.

Source: Silver device_performance
Grain: One row per (date, campaign, device)
"""

fact_device_perf = device_perf_df.alias("d").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_device.select(
        "device",
        "device_pk"
    ),
    "device",
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("campaign_pk"),
    col("device_pk"),

    # Metrics
    col("d.metrics_impressions").alias("impressions"),
    col("d.metrics_clicks").alias("clicks"),
    col("d.metrics_conversions").alias("conversions"),
    col("d.cost_inr"),
    col("d.ctr_pct"),
    col("d.cpc_inr"),
    col("d.cvr_pct"),
    col("d.cpa_inr")

)

print("fact_device_perf sample:")
display(fact_device_perf)

print(f"\nCreated fact_device_perf with {fact_device_perf.count()} daily device records")
```

```python
"""
Creates age performance fact table.

Source: Silver age_performance
Grain: One row per (date, age_range)
"""

fact_age_perf = age_perf_df.alias("a").join(
    dim_age_range.select(
        "age_range",
        "age_range_pk"
    ),
    "age_range",
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("age_range_pk"),

    # Metrics
    col("a.metrics_impressions").alias("impressions"),
    col("a.metrics_clicks").alias("clicks"),
    col("a.metrics_conversions").alias("conversions"),
    col("a.cost_inr"),
    col("a.ctr_pct"),
    col("a.cpc_inr"),
    col("a.cvr_pct"),
    col("a.cpa_inr")

)

print("fact_age_perf sample:")
display(fact_age_perf)

print(f"\nCreated fact_age_perf with {fact_age_perf.count()} daily age records")
```

```python
"""
Creates gender performance fact table.

Source: Silver gender_performance
Grain: One row per (date, gender)
"""

fact_gender_perf = gender_perf_df.alias("g").join(
    dim_gender.select(
        "gender",
        "gender_pk"
    ),
    "gender",
    "inner"
).select(

    # Keys
    col("segments_date"),
    col("gender_pk"),

    # Metrics
    col("g.metrics_impressions").alias("impressions"),
    col("g.metrics_clicks").alias("clicks"),
    col("g.metrics_conversions").alias("conversions"),
    col("g.cost_inr"),
    col("g.ctr_pct"),
    col("g.cpc_inr"),
    col("g.cvr_pct"),
    col("g.cpa_inr")

)

print("fact_gender_perf sample:")
display(fact_gender_perf)

print(f"\nCreated fact_gender_perf with {fact_gender_perf.count()} daily gender records")
```

```python
from pyspark.sql.functions import sum
```

```python
"""
Creates conversion performance fact table.

Source: Silver conversion_performance
Grain: One row per (date, campaign)

CHANGES:
- Aggregates conversion metrics across all conversion actions.
"""

conversion_summary = conversion_perf_df.groupBy(
    "segments_date",
    "campaign_id"
).agg(

    sum("metrics_conversions").alias("total_conversions"),

    sum("metrics_conversions_value").alias("total_conversion_value")

)
```

```python
fact_conversion_perf = conversion_summary.alias("c").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).select(

    # Keys
    col("segments_date"),

    col("campaign_pk"),

    # Metrics
    col("total_conversions").alias("conversions"),

    col("total_conversion_value").alias("conversion_value")

)

print("fact_conversion_perf sample:")
display(fact_conversion_perf)

print(
    f"\nCreated fact_conversion_perf with {fact_conversion_perf.count()} daily conversion records"
)
```

```python
"""
Creates conversion action performance fact table.

Source: Silver conversion_performance
Grain: One row per (date, campaign, conversion_action)
"""

fact_conversion_action_perf = conversion_perf_df.alias("cp").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_conversion_action.select(
        "conversion_action_name",
        "conversion_action_pk"
    ),
    "conversion_action_name",
    "inner"
).select(

    # Keys
    col("segments_date"),

    col("campaign_pk"),

    col("conversion_action_pk"),

    # Metrics
    col("cp.metrics_conversions").alias("conversions"),

    col("cp.metrics_conversions_value").alias("conversion_value")

)

print("fact_conversion_action_perf sample:")
display(fact_conversion_action_perf)

print(
    f"\nCreated fact_conversion_action_perf with {fact_conversion_action_perf.count()} daily conversion action records"
)
```

```python
"""
Creates ad group performance fact table.

Source: Silver ad_group_performance
Grain: One row per (date, campaign, ad_group)
"""

fact_ad_group_perf = ad_group_perf_df.alias("ag").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_ad_group.select(
        "ad_group_id",
        "ad_group_pk"
    ),
    "ad_group_id",
    "inner"
).select(

    # Keys
    col("segments_date"),

    col("campaign_pk"),

    col("ad_group_pk"),

    # Metrics
    col("ag.metrics_impressions").alias("impressions"),

    col("ag.metrics_clicks").alias("clicks"),

    col("ag.metrics_conversions").alias("conversions"),

    col("ag.cost_inr"),

    col("ag.ctr_pct"),

    col("ag.cpc_inr"),

    col("ag.cvr_pct"),

    col("ag.cpa_inr")

)

print("fact_ad_group_perf sample:")
display(fact_ad_group_perf)

print(
    f"\nCreated fact_ad_group_perf with {fact_ad_group_perf.count()} daily ad group records"
)
```

```python
"""
Creates ad performance fact table.

Source: Silver ad_performance
Grain: One row per (date, campaign, ad_group, ad)
"""

fact_ad_perf = ad_perf_df.alias("a").join(
    dim_campaign.select(
        "campaign_id",
        "campaign_pk"
    ),
    "campaign_id",
    "inner"
).join(
    dim_ad_group.select(
        "ad_group_id",
        "ad_group_pk"
    ),
    "ad_group_id",
    "inner"
).join(
    dim_ad.select(
        "ad_id",
        "ad_group_id",
        "ad_pk"
    ),
    ["ad_id", "ad_group_id"],
    "inner"
).select(

    # Keys
    col("segments_date"),

    col("campaign_pk"),

    col("ad_group_pk"),

    col("ad_pk"),

    # Metrics
    col("a.metrics_impressions").alias("impressions"),

    col("a.metrics_clicks").alias("clicks"),

    col("a.metrics_conversions").alias("conversions"),

    col("a.cost_inr"),

    col("a.ctr_pct"),

    col("a.cpc_inr"),

    col("a.cvr_pct"),

    col("a.cpa_inr")

)

print("fact_ad_perf sample:")
display(fact_ad_perf)

print(
    f"\nCreated fact_ad_perf with {fact_ad_perf.count()} daily ad records"
)
```
